```python
# Imports
import streamlit as st
import pandas as pd
```

```python
# Create a title using markdown syntax
st.write("# Python Web App")
```

```python
# Create an opening sentence using regular text_input
st.write("Hi, this is our first web app in Python! :sunglasses:")
```

```python
# Create a Pandas dataframe
df = pd.DataFrame({'col1': [1, 2], 'col2': [3, 4]})
```

```python
# Write a dataframe to output
st.write(df)
```

```python
# Create a text input using text_input()
# And save its input as input_value.
input_value = st.text_input("Enter a Message")
```

```python
# If the Display Message button is tapped,
if st.button("Display Message"):
    # Display input_value
    st.write(input_value)
```
