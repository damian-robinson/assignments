```python
# Imports
from dataclasses import dataclass
from datetime import datetime
from typing import Any
import hashlib
```

```python
# A Block class with a hash_block method
@dataclass
class Block:
    data: Any
    creator_id: int
    timestamp: str = datetime.utcnow().strftime("%H:%M:%S")

    # Create a method called hash_block
    def hash_block(self):

        # Instantiate a sha256 has variable sha
        sha = hashlib.sha256()

        # Cast & encode self.data using str() and encode()
        data = str(self.data).encode()

        # Add data to sha using update()
        sha.update(data)

        # Cast & encode creator_id using str() and encode()
        creator_id = str(self.creator_id).encode()

        # Add data to sha using update()
        sha.update(creator_id)

        # Cast & encode timestamp using str() and encode()
        timestamp = str(self.timestamp).encode()

        # Add data to sha using update()
        sha.update(timestamp)

        # Return sha as hex value using hexdigest()
        return sha.hexdigest()
```

```python
# Create a new block instance
new_block = Block(data="test", creator_id=0)
```

```python
# Calculate the block hash
block_hash = new_block.hash_block()
```

```python
# Print the block's hash
print(block_hash)
```

```python
# Create data class Counter with method update_count
@dataclass
class Counter:
    count: int = 2

    def update_count(self):
        self.count = self.count + 1
```

```python
# Create a new instance of Counter
new_counter = Counter()
```

```python
# Update the count by calling update_count
new_counter.update_count()
new_counter.update_count()
```

```python
# Print the updated value of count
print("The current count is: ", new_counter.count)
```
