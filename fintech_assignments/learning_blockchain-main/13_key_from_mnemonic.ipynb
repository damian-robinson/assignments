```python
# Imports
import os
from dotenv import load_dotenv
load_dotenv()
from mnemonic import Mnemonic
from bip44 import Wallet
from web3 import Account
```

```python
# Store MNEMONIC value from .env file as mnemonic
mnemonic = os.getenv("MNEMONIC")

# If mnemonic equals None, create a new mnemonic 
if mnemonic is None:
  mnemo = Mnemonic("english")   
  mnemonic = mnemo.generate(strength=256)

# Display value of mnemonic
display(mnemonic)
```

```python
# Instantiate new wallet using Wallet(mnemonic)
wallet = Wallet(mnemonic)
wallet
```

```python
# Instantiate wallet.derive_account("eth") as private, public
private, public = wallet.derive_account("eth")

# Display the private key
private
```

```python
# Instantiate Account.privateKeyToAccount(private) as account
account = Account.privateKeyToAccount(private)
```

```python
# Print account
print(account)
```

```python
# Print account address
print(account.address)
```

```python
# Get account private key
account.privateKey
```
