```python
# Import Web3 & EthereumTesterProvider
from web3 import Web3, EthereumTesterProvider
```

```python
# Instantiate Web3() as w3
w3 = Web3()
```

```python
# Instantiate EthereumTesterProvider() as provider
provider = EthereumTesterProvider()
```

```python
# Update Web3 instance by passing in provider
w3 = Web3(provider)
```

```python
# Get latest block info using w3.eth.get_block()
w3.eth.get_block('latest')
```

```python
# Get accounts from the blockchain using w3.eth.accounts
w3.eth.accounts
```

```python
# Set sender address
sender = '0xabBACadaBA000000000000000000000000000003'
```

```python
# Set receiver address
receiver = '0xabbAcADABa000000000000000000000000000004'
```

```python
# Set units of gas
gas = 21000
```

```python
# Convert balance using .toWei(200, 'ether')
value = w3.toWei(200, 'ether')
```

```python
# Send transaction to blockchain using w3.eth.send_transaction()
w3.eth.send_transaction({'to': receiver , 'from': sender , 'gas': gas, 'value': value})
```
