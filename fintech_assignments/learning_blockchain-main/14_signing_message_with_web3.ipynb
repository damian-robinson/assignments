```python
# Imports
import os
from dotenv import load_dotenv
load_dotenv()
from bip44 import Wallet
from eth_account import Account
from web3.auto import w3
from eth_account.messages import encode_defunct
```

```python
# Load MNEMONIC variable from .env file
mnemonic = os.getenv("MNEMONIC")
```

```python
# Check mnemonic type
type(mnemonic)
```

```python
# Instantiate Wallet(mnemonic) as wallet
wallet = Wallet(mnemonic)
```

```python
# Review wallet instance
wallet
```

```python
# Derive private & public keys using wallet.derive_account()
private, public = wallet.derive_account("eth", account=0)
```

```python
# Review private key byte string
private
```

```python
# Create account using Account.privateKeyToAccount(private)
account = Account.privateKeyToAccount(private)
```

```python
# Print account.address to access account address
print(account.address)
```

```python
# Create a message 
msg = "Zach owes David $20"
```

```python
# Encode msg as message using encode_defunct()
message = encode_defunct(text=msg)
```

```python
# Create signed message ysubg w3.eth.account.sign_message()
signed_message = w3.eth.account.sign_message(message, private_key=private)
```

```python
# Review your signed message
signed_message
```

```python
# Get associated addres using w3.eth.account.recover_message()
w3.eth.account.recover_message(message, signature=signed_message.signature)
```
