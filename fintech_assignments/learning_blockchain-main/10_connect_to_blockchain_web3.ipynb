```python
# Import Web3 & RPC Provider
from web3 import Web3, EthereumTesterProvider
```

```python
# Instantiate EthereumTesterProvider instance as provider
provider = EthereumTesterProvider()
```

```python
# Instantiate Web3 instance passing in provider as w3
w3 = Web3(provider)
```

```python
# Get w3's latest block info
w3.eth.get_block('latest')
```

```python
# Print the list of accounts on the blockchain
print(w3.eth.accounts)
```

```python
# Get account balance using .get_balance() passing in account address
w3.eth.get_balance("0xaBbACadABa000000000000000000000000000000")
```

```python
# Store account balance as balance
balance = w3.eth.get_balance("0xaBbACadABa000000000000000000000000000000")
```

```python
# Convert balance from wei to eth
w3.fromWei(balance, 'ether')
```
